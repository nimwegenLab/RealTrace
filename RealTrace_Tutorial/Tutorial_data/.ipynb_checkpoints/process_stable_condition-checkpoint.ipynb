# Analyzing Mother Machine data with stable growth conditions

We want to analyse a Mother Machine experiment of cells growing in M9+glucose expressing GFP under the controll of hi1. The experiments were conducted by Dany Chauvin and were published in Kscheschinsk and Chauvin et al 2026.

This is the "pseudo code" for the pipeline we will run

```
python3 ../RealTrace_toolbox/setup.py [input dir.] [misc. options] [set the bleaching rate] [column settings for reading the input files]

python3 ../RealTrace_toolbox/run.py [input dir.] [running mode flags for RealTrace]

python3 ../RealTrace_toolbox/plot_results.py [input dir.] [misc. options]
```


````python
import os 
import sys

import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import matplotlib as mpl
%config InlineBackend.figure_format='retina'
````

## First look at the raw data
Let us first have a look at the input data file to see which column we need to specify, which are plotted below.

````python
infile = "glucose020_hi1_20190515_rawdata.csv"
indata = pd.read_csv(infile)
indata.head()
````

````python
_, ax = plt.subplots(1,2, figsize=(10,5))

example_lane = indata[indata["lane_ID"]=="20190515.0.10"]
ax[0].scatter(example_lane["time_sec"]/60., example_lane["length_um"], s=3, alpha=0.5)
ax[0].set_xlabel("time (min)")
ax[0].set_ylabel("cell length")

ax[1].scatter(example_lane["time_sec"]/60., example_lane["gfp_nb"], s=3, alpha=0.5)
ax[1].set_xlabel("time (min)")
ax[1].set_ylabel("GFP")

plt.tight_layout()
plt.show()
````

## Setup the RealTrace run
To run RealTrace we need to write a file with initial parameter guesses and a file that specifies how the input file is read. Both can be done by using the script `setup.py`. Here, we specify the columns that we be read by RealTrace.
Also, we rescale the time to work in minutes rather than seconds. The bleaching rate was estimated to be 0.0032 1/min. We will not reoptimize this, so it will be a fixed parameter.

````python
! python3 ../RealTrace_toolbox/setup.py -d . -rescale_time 60 -beta 0.0032 -cell_tags lane_ID id -parent_tags lane_ID parent_id -time_col time_sec -length_col length_um -fp_col gfp_nb
````

## Let's look at the two files that were created by the script
The parameter file has the estimates initial values. The initial step is just 10% of the initial values (note that the bleaching rate is fixed). Also, we found that bounds (even very generous ones) helps to prevent numerial issues during the optimization. Therefore, the bounds are initial value/100 and initial values*100.

````python
! cat glucose020_hi1_20190515_rawdata_parameters.txt
````

## Run RealTrace
The script `run.py` runs RealTrace taking the files generated in the previous step. This will run around 15min. So have some tea!

Since we did not specify an output file, all output will be saved in the default, which is `glucose020_hi1_20190515_rawdata_out`.
